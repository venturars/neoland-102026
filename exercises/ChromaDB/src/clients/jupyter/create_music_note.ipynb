## 1. Prepara el notebook

Ejecuta la celda de abajo cada vez que abras el notebook o reinicies el kernel. Tiene que escribir `Ready`.

```python
import html
import sys
from pathlib import Path

from IPython.display import HTML, display

# El notebook se abre desde src/clients/jupyter: subimos dos carpetas hasta src
SRC_DIR = Path.cwd().parents[1]
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

import sdk


def show_music_notes(notes: list[dict]) -> None:
    if len(notes) == 0:
        display(HTML("<p>No music notes yet.</p>"))
        return

    rows = ""
    for note in notes:
        rows += "<tr>"
        for column in ["id", "song_link", "user_creator_id", "title", "description", "category_ids"]:
            rows += f"<td>{html.escape(str(note[column]))}</td>"
        rows += "</tr>"

    header = "<tr><th>id</th><th>song_link</th><th>user_creator_id</th><th>title</th><th>description</th><th>category_ids</th></tr>"
    display(HTML(f"<table>{header}{rows}</table>"))


print("Ready")
```

## 2. Crea la nota

Cambia los datos de la celda de abajo y ejecútala. `category_ids` usa los id del 1 al 20. Puede quedar vacía.

```python
# Cambia estos datos
user_creator_id = 1
song_link = "https://example.com/song"
title = "Example song"
description = "A short note"
category_ids = [1, 10]

try:
    note = sdk.create_music_note(user_creator_id, song_link, title, description, category_ids)
    print("Music note created:")
    show_music_notes([note])
except RuntimeError as error:
    print(f"Error: {error}")
```

## 3. Mira todas las notas

```python
show_music_notes(sdk.list_music_notes())
```
