# List users

## Qué hace

Muestra todos los usuarios guardados, ordenados por id.

## Qué pasa en la base de datos

Solo lee la tabla `users`. No cambia nada.

## Qué comando usa

`clients/terminal/listUsers.txt`. El bloque de abajo llama a la misma función, `listUsers()`.

## Anotaciones

- Si no hay usuarios, se ven los encabezados y el texto `There are no users yet.`
- Pulsa ejecutar en el bloque de abajo. No hace falta escribir código.

```python
import os
import subprocess
from pathlib import Path

# Sube de carpeta hasta encontrar la app (la carpeta que contiene logic/).
app = Path.cwd()
while app != app.parent and not (app / "logic").is_dir():
    app = app.parent
if not (app / "logic").is_dir():
    raise SystemExit("App not found.")

code = r"""
import Table from "cli-table3";
import { listUsers } from "./logic/listUsers.js";

try {
  const users = listUsers();

  const table = new Table({
    head: ["id", "name", "email", "created_at", "updated_at"],
    style: { head: ["cyan", "bold"] },
  });
  for (const user of users) {
    table.push([user.id, user.name, user.email, user.created_at, user.updated_at]);
  }
  console.log(table.toString());

  if (users.length === 0) {
    console.log("There are no users yet.");
  }
} catch (error) {
  console.error(`Error: ${error.message}`);
  process.exit(1);
}
"""

# Fuerza los colores del encabezado aunque la salida no sea una terminal.
env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

# La salida se lee en UTF-8 para que los bordes de la tabla no fallen en Windows.
completed = subprocess.run(
    ["node", "--no-warnings", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
