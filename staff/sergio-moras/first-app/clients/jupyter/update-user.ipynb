# Update user

## Qué hace

Cambia el nombre, el email o los dos de un usuario, identificado por su id, y lo muestra ya cambiado.

## Qué pasa en la base de datos

**Escribe de verdad** en `db/users.sqlite`: modifica la fila del usuario y su `updated_at`.

## Qué comando usa

`clients/terminal/update-user.txt`. El bloque de abajo llama a la misma función, `updateUser()`.

## Anotaciones

- ⚠️ Modifica un usuario real. Pon en `updateUser(...)` el id del usuario que quieras cambiar.
- El orden es id, nombre, email. `""` deja ese campo sin cambiar.
- Si el id no existe, se ve `Error: User not found: id <id>` y no cambia nada.
- Pulsa ejecutar en el bloque de abajo. No hace falta escribir código.

```python
import os
import subprocess
from pathlib import Path

# Sube de carpeta hasta encontrar la app (la carpeta que contiene logic/).
app = Path.cwd()
while app != app.parent and not (app / "logic").is_dir():
    app = app.parent
if not (app / "logic").is_dir():
    raise SystemExit("App not found.")

code = r"""
import Table from "cli-table3";
import { updateUser } from "./logic/update-user.js";

try {
  const user = updateUser(3, "Pedro Garcia", "");

  const table = new Table({
    head: ["id", "name", "email", "created_at", "updated_at"],
    style: { head: ["cyan", "bold"] },
  });
  table.push([user.id, user.name, user.email, user.created_at, user.updated_at]);
  console.log(table.toString());
} catch (error) {
  console.error(`Error: ${error.message}`);
  process.exit(1);
}
"""

# Fuerza los colores del encabezado aunque la salida no sea una terminal.
env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

# La salida se lee en UTF-8 para que los bordes de la tabla no fallen en Windows.
completed = subprocess.run(
    ["node", "--no-warnings", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
