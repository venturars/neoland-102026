# Delete user

## Qué hace

Borra un usuario, identificado por su id.

## Qué pasa en la base de datos

**Escribe de verdad** en `db/users.sqlite`: borra la fila del usuario. No se puede deshacer.

## Qué comando usa

`clients/terminal/delete-user.txt`. El bloque de abajo llama a la misma función, `deleteUser()`.

## Anotaciones

- ⚠️ Borra el usuario indicado. Pon en `deleteUser(...)` el id del usuario que quieras borrar.
- Borrar no devuelve el usuario: la tabla confirma el id borrado con una sola fila, igual que la terminal.
- Si el id no existe, se ve `Error: User not found: id <id>` y no se borra nada.
- Pulsa ejecutar en el bloque de abajo. No hace falta escribir código.

```python
import os
import subprocess
from pathlib import Path

# Sube de carpeta hasta encontrar la app (la carpeta que contiene logic/).
app = Path.cwd()
while app != app.parent and not (app / "logic").is_dir():
    app = app.parent
if not (app / "logic").is_dir():
    raise SystemExit("App not found.")

code = r"""
import Table from "cli-table3";
import { deleteUser } from "./logic/delete-user.js";

try {
  const deletedId = deleteUser(13);

  const table = new Table({
    head: ["Result"],
    style: { head: ["cyan", "bold"] },
  });
  table.push([`User deleted: id ${deletedId}`]);
  console.log(table.toString());
} catch (error) {
  console.error(`Error: ${error.message}`);
  process.exit(1);
}
"""

# Fuerza los colores del encabezado aunque la salida no sea una terminal.
env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

# La salida se lee en UTF-8 para que los bordes de la tabla no fallen en Windows.
completed = subprocess.run(
    ["node", "--no-warnings", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
