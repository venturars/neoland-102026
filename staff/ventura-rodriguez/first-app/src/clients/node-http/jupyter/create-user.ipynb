# Crear un usuario

## Qué hace

Guarda un usuario con nombre y email.

## Qué pasa en la base de datos

Inserta una fila. El id lo pone la base. El email queda en minúsculas. Las dos fechas son el momento del alta.

## Qué acción usa

`POST /users` con Ada y ada@example.com.

## Anotaciones

Si el nombre está vacío, se ve `Name is empty`. Si el email no vale, se ve `Email is invalid`. Si ese email ya está guardado, se ve `Email is already used`. Si no se llega a la base, se ve `Could not create the user`. Se abre con `pnpm jupyter-node-http` desde la app.


```python
import os
import subprocess

code = r"""
import Table from "cli-table3";

const origin = "http://127.0.0.1:3004";
const response = await fetch(origin + "/users", {
  method: "POST",
  headers: {
    "content-type": "application/json",
  },
  body: JSON.stringify({
    name: "Patricio",
    email: "patricio@example.com",
  }),
});
const result = await response.json();
if (response.ok === false) {
  console.log(result.message);
} else {
  const table = new Table({
    head: ["id", "name", "email", "createdAt", "updatedAt"],
    style: { head: ["cyan", "bold"] },
  });
  table.push([
    result.id,
    result.name,
    result.email,
    result.createdAt,
    result.updatedAt,
  ]);
  console.log(table.toString());
}
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
