# Borrar un usuario

## Qué hace

Borra el usuario de ese id.

## Qué pasa en la base de datos

Quita esa fila. No devuelve datos.

## Qué acción usa

`deleteUser(1)`.

## Anotaciones

Si no está, se ve `User not found`. Esto borra al usuario 1. Se abre con `pnpm jupyter` desde la app.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "logic").is_dir():
    app = app.parent
if not (app / "logic").is_dir():
    raise SystemExit("No encuentro la app.")

code = r"""
import Table from "cli-table3";
import { deleteUser } from "./logic/delete-user.js";

const result = deleteUser(25);

if (result === undefined) {
  const table = new Table({
    head: ["Resultado"],
    style: { head: ["cyan", "bold"] },
  });
  table.push(["No devuelve datos."]);
  console.log(table.toString());
}
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
