# Buscar usuarios por nombre

## Qué hace

Muestra los usuarios con ese nombre exacto, con las mismas mayúsculas.

## Qué pasa en la base de datos

Solo lee. No cambia filas.

## Qué acción usa

`findUserByName("Ada")`.

## Anotaciones

Si el nombre está vacío, se ve `Name is empty`. Si no hay ninguno, no hay filas y no sale `User not found`. Se abre con `pnpm jupyter` desde la app.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "logic").is_dir():
    app = app.parent
if not (app / "logic").is_dir():
    raise SystemExit("No encuentro la app.")

code = r"""
import Table from "cli-table3";
import { findUserByName } from "./logic/find-user-by-name.js";

const result = findUserByName("Ada");

const table = new Table({
  head: ["id", "name", "email", "createdAt", "updatedAt"],
  style: { head: ["cyan", "bold"] },
});

for (const user of result) {
  table.push([
    user.id,
    user.name,
    user.email,
    user.createdAt,
    user.updatedAt,
  ]);
}

console.log(table.toString());

if (result.length === 0) {
  console.log("No hay filas.");
}
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
