# Contar usuarios por género

## Qué hace

Cuenta los usuarios según el género del nombre. Lo decide llama3.2 en local. El resultado solo se muestra.

## Qué pasa en la base de datos

Solo lee los nombres. No guarda el género.

## Qué acción usa

`await usersGenderCount()`.

## Anotaciones

Hace falta llama3.2 en marcha. Sin usuarios, los tres números son 0. Si no responde, se ve `Could not classify the users`. El arranque está en [la constitución](../../../docs/CONSTITUTIONS.md#ejecución).


```python
import os
import subprocess
from pathlib import Path

# Busca la raíz de first-app aunque el notebook se abra desde src/clients/jupyter.
app = Path.cwd()
while app != app.parent and not ((app / "package.json").is_file() and (app / "src" / "logic").is_dir()):
    app = app.parent
if not ((app / "package.json").is_file() and (app / "src" / "logic").is_dir()):
    raise SystemExit("No encuentro la app.")

code = r"""
// Ejecuta la operación de la aplicación y muestra su resultado.
import Table from "cli-table3";
import { usersGenderCount } from "./src/logic/users-gender-count.js";

const result = await usersGenderCount();

const table = new Table({
  head: ["women", "men", "genderUndefined"],
  style: { head: ["cyan", "bold"] },
});

table.push([result.women, result.men, result.genderUndefined]);

console.log(table.toString());
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
