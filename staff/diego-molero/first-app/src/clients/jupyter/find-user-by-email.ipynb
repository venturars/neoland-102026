# Buscar un usuario por email

## Qué hace

Muestra el usuario de ese email.

## Qué pasa en la base de datos

Solo lee. El email se compara en minúsculas. No cambia filas.

## Qué acción usa

`findUserByEmail("ada@example.com")`.

## Anotaciones

Si el email no vale, se ve `Email is invalid`. Si no está, se ve `User not found`. El arranque está en [la constitución](../../../docs/CONSTITUTIONS.md#ejecución).


```python
import os
import subprocess
from pathlib import Path

# Busca la raíz de first-app aunque el notebook se abra desde src/clients/jupyter.
app = Path.cwd()
while app != app.parent and not ((app / "package.json").is_file() and (app / "src" / "logic").is_dir()):
    app = app.parent
if not ((app / "package.json").is_file() and (app / "src" / "logic").is_dir()):
    raise SystemExit("No encuentro la app.")

code = r"""
// Ejecuta la operación de la aplicación y muestra su resultado.
import Table from "cli-table3";
import { findUserByEmail } from "./src/logic/find-user-by-email.js";

const result = findUserByEmail("ada@example.com");

const table = new Table({
  head: ["id", "name", "email", "createdAt", "updatedAt"],
  style: { head: ["cyan", "bold"] },
});

table.push([
  result.id,
  result.name,
  result.email,
  result.createdAt,
  result.updatedAt,
]);

console.log(table.toString());
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
