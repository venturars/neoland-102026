# Listar usuarios

## Qué hace

Muestra los usuarios guardados, de menor a mayor id.

## Qué pasa en la base de datos

Solo lee. No cambia filas.

## Qué acción usa

`listUsers()`.

## Anotaciones

Si no hay usuarios, no hay filas. El arranque está en [la constitución](../../docs/CONSTITUTIONS.md#ejecución).


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "logic").is_dir():
    app = app.parent
if not (app / "logic").is_dir():
    raise SystemExit("No encuentro la app.")

code = r"""
// Ejecuta la operación de la aplicación y muestra su resultado.
import Table from "cli-table3";
import { listUsers } from "./logic/list-users.js";

const result = listUsers();

const table = new Table({
  head: ["id", "name", "email", "createdAt", "updatedAt"],
  style: { head: ["cyan", "bold"] },
});

for (const user of result) {
  table.push([
    user.id,
    user.name,
    user.email,
    user.createdAt,
    user.updatedAt,
  ]);
}

console.log(table.toString());

if (result.length === 0) {
  console.log("No hay filas.");
}
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
