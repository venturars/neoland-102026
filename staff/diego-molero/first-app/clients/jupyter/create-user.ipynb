# Crear un usuario

## Qué hace

Guarda un usuario con nombre y email.

## Qué pasa en la base de datos

Inserta una fila. El id lo pone la base. El email queda en minúsculas. Las dos fechas son el momento del alta.

## Qué acción usa

`createUser({ name: "Ada", email: "ada@example.com" })`.

## Anotaciones

Si el nombre está vacío, se ve `Name is invalid`. Si el email no vale, se ve `Email is invalid`. Si ese email ya está guardado, se ve `Email is already used`. El arranque está en [la constitución](../../docs/CONSTITUTIONS.md#ejecución).


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "logic").is_dir():
    app = app.parent
if not (app / "logic").is_dir():
    raise SystemExit("No encuentro la app.")

code = r"""
// Ejecuta la operación de la aplicación y muestra su resultado.
import Table from "cli-table3";
import { createUser } from "./logic/create-user.js";

const result = createUser({ name: "Ada", email: "ada@example.com" });

const table = new Table({
  head: ["id", "name", "email", "createdAt", "updatedAt"],
  style: { head: ["cyan", "bold"] },
});

table.push([
  result.id,
  result.name,
  result.email,
  result.createdAt,
  result.updatedAt,
]);

console.log(table.toString());
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
