# Cambiar un usuario

## Qué hace

Cambia el nombre de un usuario ya guardado. Si no se manda email, se queda el que había.

## Qué pasa en la base de datos

Actualiza esa fila y la fecha de cambio. El id no cambia.

## Qué acción usa

`updateUser(1, { name: "Ada Lovelace" })`.

## Anotaciones

Si no está, se ve `User not found`. Si el nombre está vacío, se ve `Name is invalid`. Si el email no vale, se ve `Email is invalid`. Si el email es de otro usuario, se ve `Email is already used`. El arranque está en [la constitución](../../docs/CONSTITUTIONS.md#ejecución).


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "logic").is_dir():
    app = app.parent
if not (app / "logic").is_dir():
    raise SystemExit("No encuentro la app.")

code = r"""
// Ejecuta la operación de la aplicación y muestra su resultado.
import Table from "cli-table3";
import { updateUser } from "./logic/update-user.js";

const result = updateUser(1, { name: "Ada Lovelace" });

const table = new Table({
  head: ["id", "name", "email", "createdAt", "updatedAt"],
  style: { head: ["cyan", "bold"] },
});

table.push([
  result.id,
  result.name,
  result.email,
  result.createdAt,
  result.updatedAt,
]);

console.log(table.toString());
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
