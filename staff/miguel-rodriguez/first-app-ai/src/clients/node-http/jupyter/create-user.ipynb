# Crear contacto
Inicia `pnpm server` y `pnpm node-http` antes de ejecutar la celda. Para operaciones por ID, usa un ID existente.


```python
import json
from pathlib import Path
import subprocess

# Cambia los valores para probar con tus contactos.
request = {"method": "POST", "path": "/users", "body": {"name": "Ada", "email": "ada@example.com"}}
app = Path.cwd().parents[3]
result = subprocess.run(
    ["node", str(app / "src/clients/node-http/run-request.js")],
    input=json.dumps(request), text=True, capture_output=True, cwd=app,
)
print(result.stdout, end="")
if result.stderr:
    print(result.stderr, end="")
```
