# Crear un usuario

## Qué hace

Crea un usuario con el nombre Ada y el correo ada@example.com, y muestra el usuario creado.

## Qué pasa en la base de datos

Inserta una fila. El id y las dos fechas los pone la base.

## Qué acción usa

`POST /users` con `{ "name": "Ada", "email": "ada@example.com" }`.

## Anotaciones

Si el nombre está vacío, se ve `Name is empty`. Si el correo no vale, se ve `Email is invalid`. Si el correo ya lo tiene otro usuario, se ve `Email is already used`; es lo que pasa si se ejecuta dos veces. Si no se puede guardar, se ve `Could not create the user`.

Se abre con `pnpm jupyter-node-http` desde la app.

```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "src" / "logic").is_dir():
    app = app.parent
if not (app / "src" / "logic").is_dir():
    raise SystemExit("No encuentro la app.")

code = r"""
import Table from "cli-table3";

const origin = "http://127.0.0.1:3004";
const response = await fetch(origin + "/users", {
  method: "POST",
  headers: { "Content-Type": "application/json" },
  body: JSON.stringify({ name: "Ada", email: "ada@example.com" }),
});
const result = await response.json();
if (response.ok === false) {
  console.log(result.message);
} else {
  const table = new Table({
    head: ["id", "name", "email", "createdAt", "updatedAt"],
    style: { head: ["cyan", "bold"] },
  });
  table.push([
    result.id,
    result.name,
    result.email,
    result.createdAt,
    result.updatedAt,
  ]);
  console.log(table.toString());
}
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
