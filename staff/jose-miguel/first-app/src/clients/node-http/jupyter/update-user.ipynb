# Modificar un usuario

## Qué hace

Cambia el nombre de Ada a Ada Lovelace. El correo queda igual y el id no cambia.

## Qué pasa en la base de datos

Cambia el nombre, el correo y la fecha de modificación de la fila. No crea ni borra filas.

## Qué acción usa

`PATCH /users/:id` con el id de Ada y `{ "name": "Ada Lovelace", "email": "ada@example.com" }`. El id se busca antes en `GET /users`, por el correo `ada@example.com`.

## Anotaciones

Si el id no vale, se ve `Id is invalid`. Si el nombre está vacío, se ve `Name is empty`. Si el correo no vale, se ve `Email is invalid`. Si no hay usuario con ese id, se ve `User not found`. Si el correo ya lo tiene otro usuario, se ve `Email is already used`. Si no se puede guardar, se ve `Could not update the user`.

Se abre con `pnpm jupyter-node-http` desde la app.

```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "src" / "logic").is_dir():
    app = app.parent
if not (app / "src" / "logic").is_dir():
    raise SystemExit("No encuentro la app.")

code = r"""
import Table from "cli-table3";

const origin = "http://127.0.0.1:3004";
const usersResponse = await fetch(origin + "/users");
const users = await usersResponse.json();
let adaId;
for (const user of users) {
  if (user.email === "ada@example.com") {
    adaId = user.id;
  }
}
const response = await fetch(origin + "/users/" + adaId, {
  method: "PATCH",
  headers: { "Content-Type": "application/json" },
  body: JSON.stringify({ name: "Ada Lovelace", email: "ada@example.com" }),
});
const result = await response.json();
if (response.ok === false) {
  console.log(result.message);
} else {
  const table = new Table({
    head: ["id", "name", "email", "createdAt", "updatedAt"],
    style: { head: ["cyan", "bold"] },
  });
  table.push([
    result.id,
    result.name,
    result.email,
    result.createdAt,
    result.updatedAt,
  ]);
  console.log(table.toString());
}
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
