# Conteo de usuarios por género

## Qué hace

Cuenta los usuarios por género. Clasifica sus nombres con Ollama, con el modelo `llama3.2`. No cambia nada.

## Qué pasa en la base de datos

Lee las filas. No escribe, no cambia y no borra.

## Qué acción usa

`GET /users/gender-count`.

## Anotaciones

Necesita Ollama en marcha en `http://127.0.0.1:11434` con el modelo `llama3.2` descargado. Si Ollama no responde, se ve `Could not classify the users`. Sin usuarios, la tabla muestra 0, 0 y 0 sin llamar a Ollama. Un nombre que el modelo no clasifica cuenta como `genderUndefined`.

Se abre con `pnpm jupyter-node-http` desde la app.

```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "src" / "logic").is_dir():
    app = app.parent
if not (app / "src" / "logic").is_dir():
    raise SystemExit("No encuentro la app.")

code = r"""
import Table from "cli-table3";

const origin = "http://127.0.0.1:3004";
const response = await fetch(origin + "/users/gender-count");
const result = await response.json();
if (response.ok === false) {
  console.log(result.message);
} else {
  const table = new Table({
    head: ["women", "men", "genderUndefined"],
    style: { head: ["cyan", "bold"] },
  });
  table.push([result.women, result.men, result.genderUndefined]);
  console.log(table.toString());
}
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
