# Find by name

Lists the users whose name is Ana.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "logic").is_dir():
    app = app.parent
if not (app / "logic").is_dir():
    raise SystemExit("Cannot find the app.")

code = r"""
import Table from "cli-table3";
import { db } from "./db/connection.js";
import { findUserByName } from "./logic/find-user-by-name.js";
import { createSqliteStore } from "./queries/users.js";

const store = createSqliteStore(db);
const result = await findUserByName(store, "Ana");

const table = new Table({
  head: ["id", "nombre", "email"],
  style: { head: ["cyan", "bold"] },
});
for (const user of result) {
  table.push([user.id, user.nombre, user.email]);
}
console.log(table.toString());
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
