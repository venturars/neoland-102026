# Borrar un usuario

Borra el usuario que tiene ese id.

**Cuidado:** este cuaderno borra de verdad el usuario de tu base, y no se puede deshacer. Mira antes qué usuario tiene ese id.

## La petición

`DELETE /users/:id`, sin cuerpo. Cambia `id` por el de un usuario que exista. El cuaderno de crear te da uno, y el de listar te los enseña todos.

## Qué pasa en la base

Quita esa fila.

## Las respuestas

| Estado | Qué se ve |
| --- | --- |
| 204 | Solo el estado. Si ha ido bien, la respuesta no trae nada |
| 404 | `User not found`. No hay ningún usuario con ese id |
| 500 | `Could not delete the user`. No se ha llegado a la base |

## Qué tiene que estar en marcha

Antes de ejecutarlo, lanza `pnpm start` y `pnpm node-http`, cada uno en su terminal.


```python
# Lanza programas y lee las variables del sistema.
import os
import subprocess
# Permite subir de carpeta hasta encontrar la app.
from pathlib import Path

# Busca la carpeta de la app, para que Node encuentre sus paquetes.
app = Path.cwd()
while app != app.parent and not (app / "src" / "logic").is_dir():
    app = app.parent
if not (app / "src" / "logic").is_dir():
    raise SystemExit("Cannot find the app.")

# El programa de Node que llama a la API.
code = r"""
// Dirección de la API.
const ORIGIN = "http://127.0.0.1:3004";

// El id del usuario. Cámbialo por uno que exista.
const id = 1;

// Lanza la petición.
const response = await fetch(`${ORIGIN}/users/${id}`, { method: "DELETE" });

// Muestra el método, la ruta y el estado de la respuesta.
console.log(`DELETE /users/${id}: ${response.status}`);

// Lee la respuesta. Algunas respuestas no traen nada.
const text = await response.text();
const result = text === "" ? null : JSON.parse(text);

// Si ha fallado, se ve el mensaje.
if (!response.ok) {
  console.log(result?.message ?? "");
}
"""

# Pide a Node que escriba con colores.
env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

# Lanza Node y muestra lo que escribe.
completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
