# Count by gender

Counts women, men and undefined with local llama3.2. Ollama has to be running.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "src" / "logic").is_dir():
    app = app.parent
if not (app / "src" / "logic").is_dir():
    raise SystemExit("Cannot find the app.")

code = r"""
import { db } from "./src/db/connection.js";
import { usersGenderCount } from "./src/logic/users-gender-count.js";
import { createSqliteStore } from "./src/queries/users.js";

const store = createSqliteStore(db);
const result = await usersGenderCount(store);
console.log(result);
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
