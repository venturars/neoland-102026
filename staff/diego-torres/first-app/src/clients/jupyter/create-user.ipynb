# Create a user

Saves a user with a name and an email. Open it with `pnpm run jupyter` from the app.

Start the server first with `pnpm start` in another terminal.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "src" / "logic").is_dir():
    app = app.parent
if not (app / "src" / "logic").is_dir():
    raise SystemExit("Cannot find the app.")

code = r"""
import Table from "cli-table3";
import { createUser } from "./src/logic/create-user.js";

const result = await createUser({
  name: "Luis",
  email: "luis@ejemplo.com",
});

const table = new Table({
  head: ["id", "name", "email", "createdAt", "updatedAt"],
  style: { head: ["cyan", "bold"] },
});
table.push([result.id, result.name, result.email, result.createdAt, result.updatedAt]);
console.log(table.toString());
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
