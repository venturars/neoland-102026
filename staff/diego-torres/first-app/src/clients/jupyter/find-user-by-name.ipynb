# Find by name

Lists the users whose name is Ana.

Start the server first with `pnpm start` in another terminal.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "src" / "logic").is_dir():
    app = app.parent
if not (app / "src" / "logic").is_dir():
    raise SystemExit("Cannot find the app.")

code = r"""
import Table from "cli-table3";
import { findUserByName } from "./src/logic/find-user-by-name.js";

const result = await findUserByName("Ana");

const table = new Table({
  head: ["id", "name", "email", "createdAt", "updatedAt"],
  style: { head: ["cyan", "bold"] },
});
for (const user of result) {
  table.push([user.id, user.name, user.email, user.createdAt, user.updatedAt]);
}
console.log(table.toString());
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
