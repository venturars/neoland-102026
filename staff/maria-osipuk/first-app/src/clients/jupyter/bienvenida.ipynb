# Bienvenida

**Escenario.** La aplicación guarda una lista de personas. Cada una tiene nombre, email y un número que asigna ella. Aquí se prueba, sin programar, qué se puede hacer hoy con esa lista.

**Instrucciones.** Desde la carpeta de la aplicación, `./scripts/jupyter.sh` abre esta página. El kernel, arriba a la derecha, tiene que llamarse **first-app**. Entra en un caso, cambia solo el dato que piden las instrucciones y pulsa **Run**. El resultado sale justo debajo de la celda. Si no se puede hacer, debajo aparece una frase en español. Crear, cambiar y borrar escriben en la lista real.

- [Buscar por email](buscar-usuario-por-email.ipynb)
- [Buscar por número](buscar-usuario-por-id.ipynb)
- [Actualizar el nombre o el email](actualizar-usuario.ipynb)
- [Borrar una persona](borrar-usuario.ipynb)
- [Ver la lista](listar-usuarios.ipynb)
- [Contar según el género del nombre](contar-usuarios-por-genero.ipynb)
