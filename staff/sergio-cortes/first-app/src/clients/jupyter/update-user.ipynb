# Cambiar una persona

## Qué hace

Cambia el nombre y el correo de una persona que ya está guardada.

## Qué pasa en la base de datos

Actualiza esa fila. El número sigue igual. La fecha de cambio pasa a ser el momento del cambio. La fecha de alta no se toca.

## Qué funciones del dominio usa

`updateUser({ id: 1, name: "Ada Lovelace", email: "ada@example.com" })`.

## Anotaciones

Si el número no vale, se ve `Id is invalid`. Si el nombre está vacío, se ve `Name is empty`. Si el correo no vale, se ve `Email is invalid`. Si no hay nadie con ese número, se ve `User not found`. Si el correo lo tiene otra persona, se ve `Email is already used`. Si la base no puede cambiar, se ve `Could not update the user`.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "src" / "logic").is_dir():
    app = app.parent
if not (app / "src" / "logic").is_dir():
    raise SystemExit("No encuentro la app.")

code = r"""import { updateUser } from "./src/logic/update-user.js";

const result = updateUser({
  id: 1,
  name: "Ada Lovelace",
  email: "ada@example.com",
});

process.env.FORCE_COLOR = "1";
const { default: Table } = await import("cli-table3");
await import("@colors/colors/safe.js");
const table = new Table({
  head: ["id", "name", "email", "createdAt", "updatedAt"],
  style: { head: ["cyan", "bold"] },
});
table.push([result.id, result.name, result.email, result.createdAt, result.updatedAt]);
console.log(table.toString());
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
