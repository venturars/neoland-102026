# Listar usuarios

## Qué hace

Muestra los usuarios guardados, de menor a mayor id. No cambia nada.

## Qué pasa en la base de datos

Lee las filas. No escribe, no cambia y no borra.

## Qué acción usa

`GET /users`.

## Anotaciones

Si la base no puede leer, se ve `Could not list the users`. Si no hay usuarios, la tabla deja los encabezados y la línea `No hay filas.`.

Se abre con `pnpm jupyter-node-http` desde la app.

```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "src" / "logic").is_dir():
    app = app.parent
if not (app / "src" / "logic").is_dir():
    raise SystemExit("No encuentro la app.")

code = r"""
import Table from "cli-table3";

const origin = "http://127.0.0.1:3004";
const response = await fetch(origin + "/users");
const result = await response.json();
if (response.ok === false) {
  console.log(result.message);
} else {
  const table = new Table({
    head: ["id", "name", "email", "createdAt", "updatedAt"],
    style: { head: ["cyan", "bold"] },
  });
  for (const user of result) {
    table.push([
      user.id,
      user.name,
      user.email,
      user.createdAt,
      user.updatedAt,
    ]);
  }
  console.log(table.toString());
  if (result.length === 0) {
    console.log("No hay filas.");
  }
}
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
