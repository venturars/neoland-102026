# Preparar el sitio de las personas

## Qué hace

Deja listo el sitio donde se guardan las personas. Si ya estaba listo, no borra a nadie.

## Qué pasa en la base de datos

Crea la tabla de personas si no existe. No guarda personas y no borra filas.

## Qué funciones del dominio usa

No usa una función de dominio. Usa `createUsersTable(db)`.

## Anotaciones

No borra filas. No hay un aviso de nombre, correo o persona inexistente.


```python
process.env.FORCE_COLOR = "1";

const { existsSync } = await import("node:fs");
const { dirname, join } = await import("node:path");
const { pathToFileURL } = await import("node:url");
const { default: Table } = await import("cli-table3");
await import("@colors/colors/safe.js");

let app = process.cwd();

while (!existsSync(join(app, "logic")) || !existsSync(join(app, "package.json"))) {
  const parent = dirname(app);
  if (parent === app) {
    throw new Error("No encuentro la app.");
  }
  app = parent;
}

const { db } = await import(pathToFileURL(join(app, "db/connections.js")).href);
const { createUsersTable } = await import(
  pathToFileURL(join(app, "queries/user.js")).href
);

const result = createUsersTable(db);

const table = new Table({
  head: ["lastInsertRowid", "changes"],
  style: { head: ["cyan", "bold"] },
});

table.push([result.lastInsertRowid, result.changes]);

console.log(table.toString());
```
