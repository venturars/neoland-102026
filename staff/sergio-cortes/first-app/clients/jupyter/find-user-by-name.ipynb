# Buscar por nombre

## Qué hace

Muestra las personas que se llaman exactamente así, con las mismas mayúsculas. No cambia nada.

## Qué pasa en la base de datos

Lee esas filas, de menor a mayor número. No escribe. Si no hay ninguna, no es un fallo.

## Qué funciones del dominio usa

`findUserByName("Grace")`.

## Anotaciones

Si el nombre está vacío, se ve `Name is empty`. Si la base no puede leer, se ve `Could not find the user`. Si no hay nadie con ese nombre, la tabla se queda sin filas y no se ve `User not found`.


```javascript
process.env.FORCE_COLOR = "1";

const { existsSync } = await import("node:fs");
const { dirname, join } = await import("node:path");
const { pathToFileURL } = await import("node:url");
const { default: Table } = await import("cli-table3");
await import("@colors/colors/safe.js");

let app = process.cwd();

while (!existsSync(join(app, "logic")) || !existsSync(join(app, "package.json"))) {
  const parent = dirname(app);
  if (parent === app) {
    throw new Error("No encuentro la app.");
  }
  app = parent;
}

const { findUserByName } = await import(
  pathToFileURL(join(app, "logic/find-user-by-name.js")).href
);

const result = findUserByName("Grace");

const table = new Table({
  head: ["id", "name", "email", "createdAt", "updatedAt"],
  style: { head: ["cyan", "bold"] },
});

const rows = Array.isArray(result) ? result : [result];

for (const row of rows) {
  table.push([row.id, row.name, row.email, row.createdAt, row.updatedAt]);
}

console.log(table.toString());
```
