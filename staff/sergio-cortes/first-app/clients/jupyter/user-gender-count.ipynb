# Contar por género

## Qué hace

Cuenta las personas según el género del nombre. Lo decide la IA local. No guarda el género.

## Qué pasa en la base de datos

Lee los nombres. No escribe, no cambia y no borra.

## Qué funciones del dominio usa

`await usersGenderCount()`.

## Anotaciones

Si no hay personas, los tres números son 0. Si no se puede conectar o no se pueden leer, se ve `Could not classify the users`.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "logic").is_dir():
    app = app.parent
if not (app / "logic").is_dir():
    raise SystemExit("No encuentro la app.")

code = r"""import { usersGenderCount } from "./logic/user-gender-count.js";

const result = await usersGenderCount();

process.env.FORCE_COLOR = "1";
const { default: Table } = await import("cli-table3");
await import("@colors/colors/safe.js");
const table = new Table({
  head: ["woman", "man", "genderUndefined"],
  style: { head: ["cyan", "bold"] },
});
table.push([result.woman, result.man, result.genderUndefined]);
console.log(table.toString());
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
