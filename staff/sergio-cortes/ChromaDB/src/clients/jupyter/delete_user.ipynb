## 1. Prepara el notebook

Ejecuta la celda de abajo cada vez que abras el notebook o reinicies el kernel. Tiene que escribir `Ready`.

```python
import html
import sys
from pathlib import Path

from IPython.display import HTML, display

# El notebook se abre desde src/clients/jupyter: subimos dos carpetas hasta src
SRC_DIR = Path.cwd().parents[1]
if str(SRC_DIR) not in sys.path:
    sys.path.insert(0, str(SRC_DIR))

import sdk


def show_users(users: list[dict]) -> None:
    if len(users) == 0:
        display(HTML("<p>No users yet.</p>"))
        return

    rows = ""
    for user in users:
        rows += "<tr>"
        for column in ["id", "name", "email", "age", "gender"]:
            rows += f"<td>{html.escape(str(user[column]))}</td>"
        rows += "</tr>"

    header = "<tr><th>id</th><th>name</th><th>email</th><th>age</th><th>gender</th></tr>"
    display(HTML(f"<table>{header}{rows}</table>"))


print("Ready")
```

## 2. Mira los usuarios y elige el id

```python
show_users(sdk.list_users())
```

## 3. Borra el usuario

Escribe el `id` y ejecuta la celda. El borrado no se puede deshacer.

```python
# Cambia este dato
user_id = 1

try:
    user = sdk.delete_user(user_id)
    print("User deleted:")
    show_users([user])
except RuntimeError as error:
    print(f"Error: {error}")
```

## 4. Mira todos los usuarios

```python
show_users(sdk.list_users())
```
