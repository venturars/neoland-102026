# Listar usuarios

## Qué hace

Muestra los usuarios guardados, de menor a mayor id.

## Qué pasa en la base de datos

Solo lee. No cambia filas.

## Qué ruta usa

`GET /users`.

## Anotaciones

Si no hay usuarios, el cuerpo es una lista vacía. Se abre con `pnpm node-http-jupyter` desde la app. Hace falta `pnpm server` y `pnpm node-http`.

```python
import subprocess

code = r"""
import { request } from "node:http";

const ORIGIN = "http://127.0.0.1:3004";
const originUrl = new URL(ORIGIN);
const hostname = originUrl.hostname;
const port = originUrl.port;

const result = await new Promise((resolve, reject) => {
  const req = request(
    {
      hostname,
      port,
      path: "/users",
      method: "GET",

    },
    (res) => {
      let raw = "";
      res.setEncoding("utf8");
      res.on("data", (chunk) => {
        raw += chunk;
      });
      res.on("end", () => {
        const statusCode = res.statusCode;
        resolve({ statusCode, raw });
      });
    },
  );
  req.on("error", (error) => {
    reject(error);
  });
  req.end();
});

const statusCode = result.statusCode;
const raw = result.raw;
console.log(statusCode);
if (raw !== "") {
  console.log(raw);
}
"""

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    capture_output=True,
    text=True,
    encoding="utf-8",
    errors="replace",
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
