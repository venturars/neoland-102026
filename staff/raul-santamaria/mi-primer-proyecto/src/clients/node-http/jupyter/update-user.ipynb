# Cambiar un usuario

## Qué hace

Cambia el nombre de un usuario ya guardado. Si no se manda email, se queda el que había.

## Qué pasa en la base de datos

Actualiza esa fila y la fecha de cambio. El id no cambia.

## Qué ruta usa

`PATCH /users/:id`.

## Anotaciones

Si el id no vale, se ve `User id is not valid`. Si no está, se ve `No user exists with that id`. Si el nombre está vacío, se ve `Name is empty`. Si el email no vale, se ve `Email is invalid`. Si el email es de otro usuario, se ve `Email is already used`. Se abre con `pnpm node-http-jupyter` desde la app. Hace falta `pnpm server` y `pnpm node-http`.

```python
import subprocess

code = r"""
import { request } from "node:http";

const ORIGIN = "http://127.0.0.1:3004";
const originUrl = new URL(ORIGIN);
const hostname = originUrl.hostname;
const port = originUrl.port;
const payload = JSON.stringify({ name: "Ada Lovelace" });
const payloadSize = Buffer.byteLength(payload);

const result = await new Promise((resolve, reject) => {
  const req = request(
    {
      hostname,
      port,
      path: "/users/1",
      method: "PATCH",
      headers: {
        "content-type": "application/json",
        "content-length": payloadSize,
      },
    },
    (res) => {
      let raw = "";
      res.setEncoding("utf8");
      res.on("data", (chunk) => {
        raw += chunk;
      });
      res.on("end", () => {
        const statusCode = res.statusCode;
        resolve({ statusCode, raw });
      });
    },
  );
  req.on("error", (error) => {
    reject(error);
  });
  req.write(payload);
  req.end();
});

const statusCode = result.statusCode;
const raw = result.raw;
console.log(statusCode);
if (raw !== "") {
  console.log(raw);
}
"""

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    capture_output=True,
    text=True,
    encoding="utf-8",
    errors="replace",
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
