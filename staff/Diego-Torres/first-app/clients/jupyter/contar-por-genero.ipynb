# Contar por género

Cuenta mujeres, hombres y no definidos con llama3.2 en local. Hace falta Ollama en marcha.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "logic").is_dir():
    app = app.parent
if not (app / "logic").is_dir():
    raise SystemExit("No encuentro la app.")

code = r"""
import { db } from "./db/connection.js";
import { contarUsuariosPorGenero } from "./logic/clasificador.js";
import { crearAlmacenSqlite } from "./queries/users.js";

const almacen = crearAlmacenSqlite(db);
const result = await contarUsuariosPorGenero(almacen);
console.log(result);
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
