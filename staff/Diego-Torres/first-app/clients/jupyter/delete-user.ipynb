# Eliminar un usuario

Borra el usuario con id 1 y muestra el que había.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "logic").is_dir():
    app = app.parent
if not (app / "logic").is_dir():
    raise SystemExit("No encuentro la app.")

code = r"""
import Table from "cli-table3";
import { db } from "./db/connection.js";
import { deleteUser } from "./logic/delete-user.js";
import { createSqliteStore } from "./queries/users.js";

const almacen = createSqliteStore(db);
const result = await deleteUser(almacen, "1");

const table = new Table({
  head: ["id", "nombre", "email"],
  style: { head: ["cyan", "bold"] },
});
table.push([result.id, result.nombre, result.email]);
console.log(table.toString());
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
