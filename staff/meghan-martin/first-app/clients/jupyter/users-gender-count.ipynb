# Contar por género

## Qué hace

Cuenta las personas según el género del nombre.

## Qué pasa en la base de datos

Solo lee. No guarda el género.

## Qué acción usa

`usersGenderCount()`.

## Anotaciones

Hace falta Ollama en marcha, con el modelo `nimble-v3` o el de `OLLAMA_MODEL`. Sin personas, los tres números son 0. Si no responde, `Could not classify the users`. Se abre con `npm run jupyter` desde `first-app`.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "logic").is_dir():
    app = app.parent
if not (app / "logic").is_dir():
    raise SystemExit("No encuentro la app.")

code = r"""
import Table from "cli-table3";
import { usersGenderCount } from "./logic/users-gender-count.js";

const result = await usersGenderCount();

const table = new Table({
  head: ["women", "men", "genderUndefined"],
  style: { head: ["cyan", "bold"] },
});

table.push([result.women, result.men, result.genderUndefined]);

console.log(table.toString());
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
