# Actualizar un usuario

## Qué hace

Cambia el nombre de la persona 1 a Ada Lovelace.

## Qué pasa en la base de datos

Actualiza esa fila y `updatedAt`. El id no cambia. Si no se manda email, se queda el que había.

## Qué acción usa

`updateUser(1, { name: "Ada Lovelace" })`.

## Anotaciones

`User not found`, `Name is empty`, `Email is invalid`, `Email is already used`, `Invalid id`. Se abre con `npm run jupyter` desde `first-app`.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "logic").is_dir():
    app = app.parent
if not (app / "logic").is_dir():
    raise SystemExit("No encuentro la app.")

code = r"""
import Table from "cli-table3";
import { updateUser } from "./logic/update-user.js";

const result = updateUser(1, { name: "Ada Lovelace" });

const table = new Table({
  head: ["id", "name", "email", "createdAt", "updatedAt"],
  style: { head: ["cyan", "bold"] },
});

table.push([
  result.id,
  result.name,
  result.email,
  result.createdAt,
  result.updatedAt,
]);

console.log(table.toString());
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
