# Buscar por email

## Qué hace

Muestra la persona de ese email.

## Qué pasa en la base de datos

Solo lee. El email se compara en minúsculas.

## Qué acción usa

`findUserByEmail("ada@example.com")`.

## Anotaciones

`Email is invalid`, `User not found`. Si no está, no hay tabla. Se abre con `npm run jupyter` desde `first-app`.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "logic").is_dir():
    app = app.parent
if not (app / "logic").is_dir():
    raise SystemExit("No encuentro la app.")

code = r"""
import Table from "cli-table3";
import { findUserByEmail } from "./logic/find-user-by-email.js";

const result = findUserByEmail("ada@example.com");

const table = new Table({
  head: ["id", "name", "email", "createdAt", "updatedAt"],
  style: { head: ["cyan", "bold"] },
});

table.push([
  result.id,
  result.name,
  result.email,
  result.createdAt,
  result.updatedAt,
]);

console.log(table.toString());
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
