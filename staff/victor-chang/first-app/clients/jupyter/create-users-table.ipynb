# Create the users table

## What it does

Prepares the users table in the database. It returns no data. Running it again changes nothing.

## What happens to the saved users

Nothing. If the table exists, the saved users stay the same.

## Feature it uses

`createUsersTable()`, from `queries/users.js`. It is a setup step, not a `logic/` feature.

## Failures it can show

- The table cannot be created → the database's own message.

Open it with `pnpm jupyter` from `first-app`.

```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while app != app.parent and not (app / "logic").is_dir():
    app = app.parent

code = r"""
try {
  const { default: Table } = await import("cli-table3");
  const { createUsersTable } = await import("./queries/users.js");
  const { listUsers } = await import("./logic/list-users.js");

  function printNoData() {
    const table = new Table({
      head: ["result"],
      style: { head: ["cyan", "bold"] },
    });
    table.push(["Returns no data."]);
    console.log(table.toString());
  }

  function printUsers(users) {
    const table = new Table({
      head: ["id", "name", "email", "createdAt", "updatedAt"],
      style: { head: ["cyan", "bold"] },
    });
    for (const user of users) {
      table.push([user.id, user.name, user.email, user.createdAt, user.updatedAt]);
    }
    console.log(table.toString());
  }

  // 1. Create the users table: a table that says it returns no data
  createUsersTable();
  printNoData();

  // 2. List the users: a table with the saved users
  printUsers(listUsers());

  // 3. Create it again: the same two tables, nothing changed
  createUsersTable();
  printNoData();
  printUsers(listUsers());
} catch (error) {
  console.log(error.message);
}
"""

if (app / "logic").is_dir():
    env = os.environ.copy()
    env["FORCE_COLOR"] = "1"
    env.pop("NO_COLOR", None)
    completed = subprocess.run(
        ["node", "--no-warnings", "--input-type=module", "-e", code],
        cwd=app,
        capture_output=True,
        text=True,
        env=env,
    )
    print(completed.stdout, end="")
    if completed.returncode != 0:
        print(completed.stderr, end="")
else:
    print("Cannot find first-app.")
```
