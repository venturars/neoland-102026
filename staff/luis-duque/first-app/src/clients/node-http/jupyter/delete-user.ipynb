# Elimina un usuario por su id

Elimina un usuario por su id.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
while not (app / "src" / "logic").is_dir():
    if app.parent == app:
        print("No encuentro la app.")
        raise SystemExit(1)
    app = app.parent

code = r"""import Table from "cli-table3";

const origin = "http://127.0.0.1:3004";

let listResponse;
try {
  listResponse = await fetch(origin + "/users");
} catch (error) {
  console.error(error.message);
  process.exit(1);
}

const users = await listResponse.json();
if (listResponse.ok === false) {
  console.error(users.message);
  process.exit(1);
}

const ada = users.find((user) => user.email === "ada@example.com");
if (ada === undefined) {
  console.error("No encuentro a Ada.");
  process.exit(1);
}

const id = ada.id;

let response;
try {
  response = await fetch(origin + "/users/" + id, {
    method: "DELETE",
  });
} catch (error) {
  console.error(error.message);
  process.exit(1);
}

if (response.ok === false) {
  const result = await response.json();
  console.error(result.message);
  process.exit(1);
}

const table = new Table({
  head: ["id", "name", "email"],
  style: { head: ["cyan", "bold"] },
});
console.log(table.toString());
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
