# Actualiza un usuario con el id y el nombre, el email o los dos

Actualiza un usuario con el id y el nombre, el email o los dos.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
if not (app / "src" / "logic").is_dir():
    app = app.parent.parent.parent

code = r"""import Table from "cli-table3";
import { updateUser } from "./src/logic/update-user.js";

try {
  const result = await updateUser(1, { name: "Ada Lovelace", email: "ada@example.com" });
  const table = new Table({
    head: ["id", "name", "email"],
    style: { head: ["cyan", "bold"] },
  });
  table.push([result.id, result.name, result.email]);
  console.log(table.toString());
} catch (error) {
  console.error(error.message);
  process.exit(1);
}"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
