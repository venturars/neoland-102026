# Cuenta el género de los nombres guardados

Cuenta el género de los nombres guardados.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
if not (app / "logic").is_dir():
    app = app.parent.parent

code = r"""import Table from "cli-table3";
import { usersGender } from "./src/logic/users-gender.js";

try {
  const result = await usersGender();
  const table = new Table({
    head: ["women", "man", "undefined"],
    style: { head: ["cyan", "bold"] },
  });
  table.push([result.women, result.man, result.undefined]);
  console.log(table.toString());
} catch (error) {
  console.error(error.message);
  process.exit(1);
}"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
