# Busca un usuario por nombre

Busca un usuario por nombre.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd()
if not (app / "logic").is_dir():
    app = app.parent.parent

code = r"""import Table from "cli-table3";
import { findUserByName } from "./src/logic/find-user-by-name.js";

try {
  const result = await findUserByName("Carmen");
  const rows = Array.isArray(result) ? result : [result];
  const table = new Table({
    head: ["id", "name", "email"],
    style: { head: ["cyan", "bold"] },
  });
  for (const user of rows) {
    table.push([user.id, user.name, user.email]);
  }
  console.log(table.toString());
} catch (error) {
  console.error(error.message);
  process.exit(1);
}"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
