# Listar usuarios

## Qué hace

Muestra los usuarios guardados, de menor a mayor id.

## Qué pasa en la base de datos

Solo lee. No cambia filas.

## Qué acción usa

La acción de dominio es `listUsers`. El bloque llama a `findAll()`.

## Anotaciones

Si no hay usuarios, no hay filas. No es un fallo. Se abre con `pnpm jupyter` desde la app.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd().parent.parent.parent

code = r"""
import "@colors/colors";
import Table from "cli-table3";
import { findAll } from "./src/logic/list-users.js";

try {
  const result = findAll();

  const table = new Table({
    head: ["id", "name", "email", "createdAt", "updatedAt"],
    style: { head: ["cyan", "bold"] },
  });

  if (Array.isArray(result)) {
    for (const user of result) {
      table.push([
        user.id,
        user.name,
        user.email,
        user.createdAt,
        user.updatedAt,
      ]);
    }
  } else if (result) {
    table.push([
      result.id,
      result.name,
      result.email,
      result.createdAt,
      result.updatedAt,
    ]);
  }

  console.log(table.toString());
} catch (error) {
  console.error(error.message);
  process.exit(1);
}
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
if completed.returncode != 0:
    print(completed.stderr or "", end="")
else:
    print(completed.stdout or "", end="")
```
