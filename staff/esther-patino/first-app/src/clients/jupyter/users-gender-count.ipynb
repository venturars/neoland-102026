# Contar por sexo del nombre

## Qué hace

Cuenta los usuarios según el sexo que sugiere el nombre. Lo decide la IA local. El resultado solo se muestra. No es el sexo verificado de la persona.

## Qué pasa en la base de datos

Solo lee. No guarda el sexo ni otro campo.

## Qué acción usa

La acción de dominio y el bloque llaman a `await usersGenderCount()`.

## Anotaciones

Hace falta la IA local en marcha. Sin usuarios, los tres números son 0. Si no responde, `Could not classify the users`. Se abre con `pnpm jupyter` desde la app.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd().parent.parent.parent

code = r"""
import "@colors/colors";
import Table from "cli-table3";
import { usersGenderCount } from "./src/logic/users-gender-count.js";

try {
  const result = await usersGenderCount();

  const table = new Table({
    head: ["women", "men", "undefinedUser"],
    style: { head: ["cyan", "bold"] },
  });
  table.push([result.women, result.men, result.undefinedUser]);
  console.log(table.toString());
} catch (error) {
  console.error(error.message);
  process.exit(1);
}
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
if completed.returncode != 0:
    print(completed.stderr or "", end="")
else:
    print(completed.stdout or "", end="")
```
