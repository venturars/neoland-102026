# Borrar un usuario

## Qué hace

Borra el usuario de ese id. Quita la fila.

## Qué pasa en la base de datos

Borra esa fila. No guarda datos nuevos.

## Qué acción usa

La acción de dominio es `deleteUser`. El bloque llama a `removeUser(1)`.

## Anotaciones

Esto borra al usuario 1. Esta llamada no muestra `User not found` si el id no está. `deleteUser` sí. `deleteUser` no devuelve el usuario. Esta llamada muestra cuántas filas quitó. Se abre con `pnpm jupyter` desde la app.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd().parent.parent.parent

code = r"""
import "@colors/colors";
import Table from "cli-table3";
import { removeUser } from "./src/logic/delete-user.js";

try {
  const result = removeUser(1);

  const table = new Table({
    head: ["changes"],
    style: { head: ["cyan", "bold"] },
  });
  table.push([result.changes]);
  console.log(table.toString());
} catch (error) {
  console.error(error.message);
  process.exit(1);
}
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
if completed.returncode != 0:
    print(completed.stderr or "", end="")
else:
    print(completed.stdout or "", end="")
```
