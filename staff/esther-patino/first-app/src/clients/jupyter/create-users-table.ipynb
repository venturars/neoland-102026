# Crear la tabla de usuarios

## Qué hace

Abre la base. Si la tabla de usuarios no existe, se crea. Luego cierra la base. No guarda usuarios.

## Qué pasa en la base de datos

Crea la tabla `users` si no existe. No guarda ni borra usuarios. No sustituye el archivo.

## Qué acción usa

No hay acción de dominio. El bloque abre `db` y lo cierra, igual que el cliente de terminal.

## Anotaciones

No borra filas y no devuelve un usuario. Se abre con `pnpm jupyter` desde la app.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd().parent.parent.parent

code = r"""
import "@colors/colors";
import Table from "cli-table3";
import { db } from "./src/db/connection.js";

try {
  db.close();

  const table = new Table({
    head: ["Resultado"],
    style: { head: ["cyan", "bold"] },
  });
  table.push(["No devuelve datos."]);
  console.log(table.toString());
} catch (error) {
  console.error(error.message);
  process.exit(1);
}
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
if completed.returncode != 0:
    print(completed.stderr or "", end="")
else:
    print(completed.stdout or "", end="")
```
