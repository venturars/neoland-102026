# Listar usuarios

## Qué hace

Muestra los usuarios guardados, de menor a mayor id.

## Qué pasa en la base de datos

Solo lee. No cambia filas.

## Qué acción usa

`GET /users`.

## Anotaciones

Si no hay usuarios, no hay filas. Si no se llega a la base, se ve `Could not list the users`. Se abre con `pnpm jupyter-node-http` desde la app.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd().parent.parent.parent.parent

code = r"""
import "@colors/colors";
import Table from "cli-table3";

const origin = "http://127.0.0.1:3004";
const response = await fetch(origin + "/users");
const result = await response.json();
if (response.ok === false) {
  console.log(result.message);
} else {
  const table = new Table({
    head: ["id", "name", "email", "createdAt", "updatedAt"],
    style: { head: ["cyan", "bold"] },
  });
  for (const user of result) {
    table.push([
      user.id,
      user.name,
      user.email,
      user.createdAt,
      user.updatedAt,
    ]);
  }
  console.log(table.toString());
  if (result.length === 0) {
    console.log("No hay filas.");
  }
}
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    capture_output=True,
    text=True,
    env=env,
    cwd=app,
)
print(completed.stdout, end="")
if completed.returncode != 0:
    print(completed.stderr, end="")
```
