# Cambiar un usuario

## Qué hace

Cambia el nombre y el email de un usuario ya guardado. La base actualiza `updatedAt`.

## Qué pasa en la base de datos

Actualiza esa fila. El id no cambia.

## Qué acción usa

La acción de dominio es `updateUser`. El bloque llama a `updateUserQuery({ id: 1, name: "Ada Lovelace", email: "ada@example.com" })`.

## Anotaciones

Esta llamada no comprueba si el usuario existe ni el email. `updateUser` sí, y puede mostrar `User not found`, `Name is empty`, `Email is invalid` o `Email is already used`. Esto cambia al usuario 1. Se abre con `pnpm jupyter` desde la app.


```python
import os
import subprocess
from pathlib import Path

app = Path.cwd().parent.parent

code = r"""
import "@colors/colors";
import Table from "cli-table3";
import { updateUserQuery } from "./logic/update-user.js";

try {
  const result = updateUserQuery({
    id: 1,
    name: "Ada Lovelace",
    email: "ada@example.com",
  });

  const table = new Table({
    head: ["id", "name", "email", "createdAt", "updatedAt"],
    style: { head: ["cyan", "bold"] },
  });

  if (Array.isArray(result)) {
    for (const user of result) {
      table.push([
        user.id,
        user.name,
        user.email,
        user.createdAt,
        user.updatedAt,
      ]);
    }
  } else if (result) {
    table.push([
      result.id,
      result.name,
      result.email,
      result.createdAt,
      result.updatedAt,
    ]);
  }

  console.log(table.toString());
} catch (error) {
  console.error(error.message);
  process.exit(1);
}
"""

env = os.environ.copy()
env["FORCE_COLOR"] = "1"
env.pop("NO_COLOR", None)

completed = subprocess.run(
    ["node", "--input-type=module", "-e", code],
    cwd=app,
    capture_output=True,
    text=True,
    encoding="utf-8",
    env=env,
)
if completed.returncode != 0:
    print(completed.stderr or "", end="")
else:
    print(completed.stdout or "", end="")
```
