# Buscar un usuario por email

Encuentra a una persona por su correo. No distingue mayúsculas ni espacios sobrantes. **Qué cambia:** nada, solo lee (⚠️ **usa la base de datos real** (`db/users.sqlite`)).

| Dato | Significado | ¿Obligatorio? | Ejemplo válido |
|---|---|---|---|
| `email` | Algo antes y después de `@`, menos de 64 caracteres | Sí | `lin@example.com`, `  LIN@Example.com ` |

**Prueba:** cambia el email marcado con ✏️. Pulsa **▶ Run** (o `Shift + Enter`) en la celda de código. **Cómo reconocer que ha funcionado:** sale una tabla con el usuario.

**Errores posibles y cómo corregirlos:**

- `The email is not valid.`: el email no tiene algo antes y después de la `@`. Escríbelo completo.
- `No user was found.`: el email es válido, pero nadie lo tiene. Comprueba que está bien escrito o míralo en *Listar usuarios*.

[← Volver al índice](00-welcome-and-index.ipynb)

```python
import json
import os
import subprocess

# ✏️ CAMBIA AQUÍ
email = "lin@example.com"

js = """
import Table from "cli-table3";
import { findUserByEmail } from "./logic/find-user-by-email.js";

const { email } = JSON.parse(process.argv[1]);

const columns = ["id", "name", "email", "createdAt", "updatedAt"];
const table = new Table({ head: columns, style: { head: ["cyan", "bold"] } });

try {
  const user = findUserByEmail({ email });
  table.push(columns.map((column) => user[column]));
  console.log(table.toString());
} catch (error) {
  console.log(error.message);
}
"""

# Lanza Node desde la carpeta first-app (dos carpetas más arriba) y le pasa los datos como JSON
result = subprocess.run(
    ["node", "--disable-warning=ExperimentalWarning", "--input-type=module", "-e", js, json.dumps({"email": email})],
    cwd="../..",
    env={**os.environ, "FORCE_COLOR": "1"},  # conserva el color de los encabezados
    capture_output=True,
    text=True,
)
print(result.stdout or result.stderr)
```
