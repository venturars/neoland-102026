# Crear un usuario

Da de alta a una persona: guarda su nombre y email, y el sistema le asigna un id y la fecha de alta. **Qué cambia:** ⚠️ **usa la base de datos real** (`db/users.sqlite`) y **añade un usuario**.

| Dato | Significado | ¿Obligatorio? | Ejemplo válido |
|---|---|---|---|
| `name` | Nombre de la persona | Sí, no vacío | `Sofía Martín` |
| `email` | Algo antes y después de `@`, menos de 64 caracteres. Se guarda en minúsculas y no puede repetirse | Sí | `sofia@example.com` |

**Prueba:** cambia los dos valores marcados con ✏️. Pulsa **▶ Run** (o `Shift + Enter`) en la celda de código. **Cómo reconocer que ha funcionado:** sale una tabla con el usuario nuevo, su `id` y las fechas. Luego puedes verlo en *Listar usuarios*.

**Errores posibles y cómo corregirlos:**

- `The name is missing.`: el nombre está vacío o solo tiene espacios. Escribe un nombre.
- `The email is not valid.`: el email no tiene algo antes y después de la `@`, o tiene 64 caracteres o más. Escríbelo completo.
- `This email is already used.`: otro usuario ya tiene ese email (por ejemplo, si repites la prueba sin cambiarlo). Usa un email distinto.

[← Volver al índice](00-welcome-and-index.ipynb)

```python
import json
import os
import subprocess

# ✏️ CAMBIA AQUÍ
name = "Sofía Martín"
email = "sofia@example.com"

js = """
import Table from "cli-table3";
import { createUser } from "./logic/create-user.js";

const { name, email } = JSON.parse(process.argv[1]);

const columns = ["id", "name", "email", "createdAt", "updatedAt"];
const table = new Table({ head: columns, style: { head: ["cyan", "bold"] } });

try {
  const user = createUser({ name, email });
  table.push(columns.map((column) => user[column]));
  console.log(table.toString());
} catch (error) {
  console.log(error.message);
}
"""

# Lanza Node desde la carpeta first-app (dos carpetas más arriba) y le pasa los datos como JSON
result = subprocess.run(
    ["node", "--disable-warning=ExperimentalWarning", "--input-type=module", "-e", js, json.dumps({"name": name, "email": email})],
    cwd="../..",
    env={**os.environ, "FORCE_COLOR": "1"},  # conserva el color de los encabezados
    capture_output=True,
    text=True,
)
print(result.stdout or result.stderr)
```
