# Actualizar un usuario

Cambia el nombre, el email o ambos de un usuario que ya existe. Lo que no envíes se queda como estaba. **Qué cambia:** ⚠️ **usa la base de datos real** (`db/users.sqlite`) y **modifica un usuario** y su fecha `updatedAt`. No hay forma de deshacerlo.

| Dato | Significado | ¿Obligatorio? | Ejemplo válido |
|---|---|---|---|
| `id` | Número del usuario (míralo en *Listar usuarios*) | Sí | `3` |
| `name` | Nombre nuevo. `None` para no cambiarlo | No, pero hace falta `name` o `email` | `"Lin Chen"` |
| `email` | Email nuevo, que no use otro usuario. `None` para no cambiarlo | No, pero hace falta `name` o `email` | `"lin.chen@example.com"` |

**Prueba:** el `id` empieza en `0`, que no existe: si pulsas **▶ Run** sin cambiar nada saldrá `No user was found.` y no se modifica a nadie. Cambia el `id` por uno real de la lista y el `name` o el `email` marcados con ✏️. Pulsa **▶ Run** (o `Shift + Enter`) en la celda de código. **Cómo reconocer que ha funcionado:** sale una tabla con el usuario y sus datos nuevos.

**Errores posibles y cómo corregirlos:**

- `No user was found.`: ningún usuario tiene ese id. Usa uno de *Listar usuarios*.
- `There is nothing to update.`: `name` y `email` están en `None`. Indica al menos uno.
- `The name is missing.`: el nombre nuevo está vacío. Escribe un nombre, o déjalo en `None`.
- `The email is not valid.`: el email nuevo no tiene algo antes y después de la `@`. Escríbelo completo.
- `This email is already used.`: otro usuario ya tiene ese email. Elige uno distinto.

```python
import json
import os
import subprocess

# ✏️ CAMBIA AQUÍ (id 0 no existe: no cambia a nadie)
id = 0
name = "Lin Chen"  # o None para no cambiarlo
email = None  # o un email nuevo; None para no cambiarlo

# Solo se envían los campos que se quieren cambiar
changes = {"id": id, "name": name, "email": email}
changes = {key: value for key, value in changes.items() if value is not None}

js = """
import Table from "cli-table3";
import { updateUser } from "./logic/update-user.js";

const { id, name, email } = JSON.parse(process.argv[1]);

const columns = ["id", "name", "email", "createdAt", "updatedAt"];
const table = new Table({ head: columns, style: { head: ["cyan", "bold"] } });

try {
  const user = updateUser({ id, name, email });
  table.push(columns.map((column) => user[column]));
  console.log(table.toString());
} catch (error) {
  console.log(error.message);
}
"""

# Lanza Node desde la carpeta first-app (dos carpetas más arriba) y le pasa los datos como JSON
result = subprocess.run(
    ["node", "--disable-warning=ExperimentalWarning", "--input-type=module", "-e", js, json.dumps(changes)],
    cwd="../..",
    env={**os.environ, "FORCE_COLOR": "1"},  # conserva el color de los encabezados
    capture_output=True,
    text=True,
)
print(result.stdout or result.stderr)
```
