# Eliminar un usuario

Da de baja a una persona. **Qué cambia:** ⚠️ **usa la base de datos real** (`src/db/users.sqlite`) y **borra un usuario**. No hay papelera ni forma de deshacerlo.

| Dato | Significado | ¿Obligatorio? | Ejemplo válido |
|---|---|---|---|
| `id` | Número del usuario que se borra (míralo en *Listar usuarios*) | Sí | `11` |

**Prueba:** el `id` empieza en `0`, que no existe: si pulsas **▶ Run** sin cambiar nada saldrá `No user was found.` y no se borra a nadie. Cámbialo por el id de un usuario que quieras borrar (por ejemplo, uno creado en *Crear un usuario*). Pulsa **▶ Run** (o `Shift + Enter`) en la celda de código. **Cómo reconocer que ha funcionado:** sale `Usuario eliminado.` y, al ejecutar *Listar usuarios*, ya no aparece.

**Error posible y cómo corregirlo:** `No user was found.` significa que ningún usuario tiene ese id (por ejemplo, si ya lo habías borrado). Usa un id que aparezca en *Listar usuarios*.

[← Volver al índice](00-welcome-and-index.ipynb)

```python
import json
import os
import subprocess

# ✏️ CAMBIA AQUÍ (id 0 no existe: no borra a nadie)
id = 0

js = """
import { deleteUser } from "./src/logic/delete-user.js";

const { id } = JSON.parse(process.argv[1]);

try {
  deleteUser({ id });
  console.log("Usuario eliminado.");
} catch (error) {
  console.log(error.message);
}
"""

# Lanza Node desde la carpeta first-app (tres carpetas más arriba) y le pasa los datos como JSON
result = subprocess.run(
    ["node", "--disable-warning=ExperimentalWarning", "--input-type=module", "-e", js, json.dumps({"id": id})],
    cwd="../../..",
    env={**os.environ, "FORCE_COLOR": "1"},  # conserva el color de los encabezados
    capture_output=True,
    text=True,
)
print(result.stdout or result.stderr)
```
