# Listar usuarios

Muestra **todos los usuarios guardados**, ordenados por id. **Qué cambia:** nada, solo lee (⚠️ **usa la base de datos real** (`src/db/users.sqlite`)). No recibe datos de entrada.

Pulsa **▶ Run** (o `Shift + Enter`) en la celda de código. **Cómo reconocer que ha funcionado:** sale una tabla con una fila por usuario (`id`, `name`, `email`, `createdAt`, `updatedAt`). Si no hubiera ninguno, solo se ven los encabezados.

**Error posible:** `The database failed.` si SQLite falla. Cierra cualquier otro programa que tenga abierto `src/db/users.sqlite` y repite.

[← Volver al índice](00-welcome-and-index.ipynb)

```python
import json
import os
import subprocess

js = """
import Table from "cli-table3";
import { listUsers } from "./src/logic/list-users.js";

const columns = ["id", "name", "email", "createdAt", "updatedAt"];
const table = new Table({ head: columns, style: { head: ["cyan", "bold"] } });

try {
  for (const user of listUsers()) {
    table.push(columns.map((column) => user[column]));
  }
  console.log(table.toString());
} catch (error) {
  console.log(error.message);
}
"""

# Lanza Node desde la carpeta first-app (tres carpetas más arriba) y le pasa los datos como JSON
result = subprocess.run(
    ["node", "--disable-warning=ExperimentalWarning", "--input-type=module", "-e", js, json.dumps({})],
    cwd="../../..",
    env={**os.environ, "FORCE_COLOR": "1"},  # conserva el color de los encabezados
    capture_output=True,
    text=True,
)
print(result.stdout or result.stderr)
```
