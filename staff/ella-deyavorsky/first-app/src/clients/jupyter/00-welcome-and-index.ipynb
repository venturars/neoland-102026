# Gestión de usuarios: prueba de concepto

La aplicación **gestiona usuarios**: cada uno tiene nombre, email único y fechas de alta y modificación. Estos cuadernos te dejan probar cada función sin programar: lee el texto, cambia (si quieres) los valores marcados con ✏️ y pulsa **▶ Run** (o `Shift + Enter`).

> ⚠️ Los cuadernos usan la **base de datos real** (`src/db/users.sqlite`). Crear, actualizar y eliminar cambian usuarios de verdad, y no hay botón para deshacerlo. Cada cuaderno avisa de qué cambia.

| Cuaderno | Qué permite hacer | ¿Cambia datos? |
|---|---|---|
| [Crear un usuario](create-user.ipynb) | Dar de alta a una persona | Sí, añade |
| [Listar usuarios](list-users.ipynb) | Ver todos los usuarios | No |
| [Buscar por id](find-user-by-id.ipynb) | Encontrar a alguien por su número | No |
| [Buscar por email](find-user-by-email.ipynb) | Encontrar a alguien por su correo | No |
| [Actualizar un usuario](update-user.ipynb) | Cambiar nombre o email | Sí, modifica |
| [Eliminar un usuario](delete-user.ipynb) | Dar de baja a alguien | Sí, borra |
| [Contar por género](count-users-by-gender.ipynb) | Cuántos hombres y mujeres hay (usa IA local con Ollama) | No |

**Paso técnico (no es una acción de usuario):** [Preparar la base de datos](prepare-database.ipynb) crea la tabla donde se guardan los usuarios, una sola vez al montar el sistema. Se incluye para entender de dónde sale la tabla.

El *kernel* debe ser **Python 3** (arriba a la derecha). Si algo se queda raro: **Kernel → Restart Kernel and Run All Cells**.
