# Buscar un usuario por id

Encuentra a una persona por su número (id). **Qué cambia:** nada, solo lee (⚠️ **usa la base de datos real** (`src/db/users.sqlite`)).

| Dato | Significado | ¿Obligatorio? | Ejemplo válido |
|---|---|---|---|
| `id` | Número único del usuario (míralo en *Listar usuarios*) | Sí | `3` |

**Prueba:** cambia el id marcado con ✏️. Pulsa **▶ Run** (o `Shift + Enter`) en la celda de código. **Cómo reconocer que ha funcionado:** sale una tabla con el usuario.

**Error posible y cómo corregirlo:** `No user was found.` significa que ningún usuario tiene ese id. Usa un id que aparezca en *Listar usuarios*.

[← Volver al índice](00-welcome-and-index.ipynb)

```python
import json
import os
import subprocess

# ✏️ CAMBIA AQUÍ
id = 3

js = """
import Table from "cli-table3";
import { findUserById } from "./src/logic/find-user-by-id.js";

const { id } = JSON.parse(process.argv[1]);

const columns = ["id", "name", "email", "createdAt", "updatedAt"];
const table = new Table({ head: columns, style: { head: ["cyan", "bold"] } });

try {
  const user = findUserById({ id });
  table.push(columns.map((column) => user[column]));
  console.log(table.toString());
} catch (error) {
  console.log(error.message);
}
"""

# Lanza Node desde la carpeta first-app (tres carpetas más arriba) y le pasa los datos como JSON
result = subprocess.run(
    ["node", "--disable-warning=ExperimentalWarning", "--input-type=module", "-e", js, json.dumps({"id": id})],
    cwd="../../..",
    env={**os.environ, "FORCE_COLOR": "1"},  # conserva el color de los encabezados
    capture_output=True,
    text=True,
)
print(result.stdout or result.stderr)
```
