# Contar usuarios por género

Cuenta cuántos usuarios son hombres, mujeres o no se pueden determinar, a partir del nombre. El género no se guarda: se calcula al momento. **Qué cambia:** nada, solo lee (⚠️ **usa la base de datos real** (`src/db/users.sqlite`)). No recibe datos de entrada.

**Antes de empezar:** Ollama debe estar en marcha con el modelo `llama3.2` (la app Ollama abierta, o `ollama serve`; y `ollama pull llama3.2` la primera vez). La IA corre en tu ordenador: ningún dato sale de él.

Pulsa **▶ Run** (o `Shift + Enter`) en la celda de código. Puede tardar unos segundos. **Cómo reconocer que ha funcionado:** sale una tabla con tres números, `male`, `female` y `undefined` (no determinado), que suman los usuarios guardados. Los decide un modelo, así que **pueden variar entre ejecuciones y no son infalibles**.

**Errores posibles y cómo corregirlos:**

- `The classifier failed.`: Ollama no responde o falta el modelo. Abre Ollama (o ejecuta `ollama serve`), comprueba con `ollama list` que está `llama3.2` y repite.
- `The database failed.`: SQLite falla. Cierra cualquier otro programa que tenga abierto `src/db/users.sqlite` y repite.

```python
import json
import os
import subprocess

js = """
import Table from "cli-table3";
import { countUsersByGender } from "./src/logic/count-users-by-gender.js";

try {
  const count = await countUsersByGender();
  const table = new Table({
    head: ["male", "female", "undefined"],
    style: { head: ["cyan", "bold"] },
  });
  table.push([count.male, count.female, count.undefined]);
  console.log(table.toString());
} catch (error) {
  console.log(error.message);
}
"""

# Lanza Node desde la carpeta first-app (tres carpetas más arriba) y le pasa los datos como JSON
result = subprocess.run(
    ["node", "--disable-warning=ExperimentalWarning", "--input-type=module", "-e", js, json.dumps({})],
    cwd="../../..",
    env={**os.environ, "FORCE_COLOR": "1"},  # conserva el color de los encabezados
    capture_output=True,
    text=True,
)
print(result.stdout or result.stderr)
```
