# Paso técnico: preparar la base de datos

> **No es una acción de usuario.** Es una operación interna que se hace una vez al montar el sistema.

Crea la tabla `users` (id, nombre, email y fechas) **si no existe**. Si ya existe no hace nada y **no borra usuarios**, así que es seguro ejecutarla. ⚠️ **usa la base de datos real** (`src/db/users.sqlite`). No recibe datos de entrada.

Pulsa **▶ Run** (o `Shift + Enter`) en la celda de código. **Cómo reconocer que ha funcionado:** sale una tabla con una fila, `users`.

**Error posible:** `The database failed.` si SQLite falla. Cierra cualquier otro programa que tenga abierto `src/db/users.sqlite` y repite.

[← Volver al índice](00-welcome-and-index.ipynb)

```python
import json
import os
import subprocess

js = """
import Table from "cli-table3";
import { db } from "./src/db/connection.js";
import { createUsersTable } from "./src/queries/users.js";

try {
  createUsersTable();
  const tables = db
    .prepare("SELECT name FROM sqlite_master WHERE type = :type")
    .all({ type: "table" });

  const table = new Table({ head: ["name"], style: { head: ["cyan", "bold"] } });
  for (const row of tables) table.push([row.name]);
  console.log(table.toString());
} catch (error) {
  console.log(error.message);
}
"""

# Lanza Node desde la carpeta first-app (tres carpetas más arriba) y le pasa los datos como JSON
result = subprocess.run(
    ["node", "--disable-warning=ExperimentalWarning", "--input-type=module", "-e", js, json.dumps({})],
    cwd="../../..",
    env={**os.environ, "FORCE_COLOR": "1"},  # conserva el color de los encabezados
    capture_output=True,
    text=True,
)
print(result.stdout or result.stderr)
```
