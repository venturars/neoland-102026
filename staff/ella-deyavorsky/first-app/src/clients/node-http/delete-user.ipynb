# Eliminar un usuario

Da de baja a una persona.

**Qué pasa en la base:** ⚠️ **usa la base de datos real** (`src/db/users.sqlite`) y **borra un usuario**. No hay papelera ni forma de deshacerlo.

**Acción que usa:** `DELETE http://127.0.0.1:3004/users/:id`. Responde `204`, sin cuerpo.

| Dato | Significado | ¿Obligatorio? | Ejemplo válido |
|---|---|---|---|
| `id` | Número del usuario que se borra (míralo en *Listar usuarios*) | Sí | `11` |

**Anotaciones:** necesitas `pnpm start` y `pnpm node-http` en marcha. El `id` empieza en `0`, que no existe: si pulsas **▶ Run** sin cambiar nada saldrá `No user was found.` y no se borra a nadie. Cámbialo por el id de un usuario que quieras borrar (por ejemplo, uno creado en *Crear un usuario*) y pulsa **▶ Run** (o `Shift + Enter`). **Cómo reconocer que ha funcionado:** sale `The user was deleted.` y, al ejecutar *Listar usuarios*, ya no aparece. No hay datos que mostrar, así que no sale tabla.

- `No user was found.` (404): ningún usuario tiene ese id (por ejemplo, si ya lo habías borrado). Usa uno de *Listar usuarios*.
- **Con la base parada:** para el programa de la base de datos (`pnpm start`) y vuelve a pulsar **▶ Run**: sale `The database failed.` y ninguna tabla.
- Si sale `The HTTP client is not running. Start it with pnpm node-http.`, arranca el cliente en otra terminal (con `pnpm start` ya en marcha) y repite.

[← Anterior: Actualizar un usuario](update-user.ipynb) · [Volver al primero: Crear un usuario](create-user.ipynb)

```python
import json
import urllib.error
import urllib.request

# ✏️ CAMBIA AQUÍ (id 0 no existe: no borra a nadie)
id = 0

URL = "http://127.0.0.1:3004/users/" + str(id)
request = urllib.request.Request(URL, method="DELETE")

# Envía la petición al cliente. Una respuesta de error (400, 404, 500) también trae un texto
status = None
text = ""
try:
    with urllib.request.urlopen(request) as answer:
        status = answer.status
        text = answer.read().decode()
except urllib.error.HTTPError as error:
    status = error.code
    text = error.read().decode()
except urllib.error.URLError:
    pass

if status is None:
    print("The HTTP client is not running. Start it with pnpm node-http.")
elif status >= 400:
    # Un fallo: solo el mensaje
    print(json.loads(text)["error"])
else:
    # 204: se borró y no hay cuerpo
    print("The user was deleted.")
```
